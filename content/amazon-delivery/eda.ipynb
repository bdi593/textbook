# Exploratory Data Analysis


## Why EDA Matters before Building a Predictive Model

Exploratory Data Analysis (EDA) is the process of understanding the data beyond just reading it. It helps us identify:

- the types of features (numeric, categorical),
- data quality issues such as missing values, duplicates, or constant columns,
- distributions and shapes of variables,
- outliers and skewness that may distort modeling,
- relationships and correlations between predictors and the target, and
- unexpected patterns or anomalies that might affect modeling.

This phase sets the foundation for building reliable predictive models.


## Understanding Feature Types

In our dataset:

- **Numeric** features include weather measurements (e.g., `temperature`, `snow_depth`) and the operational variable `total_items` (the number of items on the delivery truck).
- `zipcode` is stored as an integer, but it is an identifier, not a quantity: ZIP code 90000 is not "more" than ZIP code 10000. We still look at its distribution below, and the next notebook turns it into a region feature.
- The only **categorical** variable is the target, `classification_ontime`.

Understanding these types helps us choose the right summaries and visualizations.

## Numeric Features Distribution

We draw one histogram per numeric feature in a single faceted figure. To keep the notebook small, we bin each feature with `np.histogram` first and then draw the bin counts as bars. The chart looks the same as `px.histogram`, but it stores 40 counts per feature instead of 45,000 raw values. Each panel has its own x and y scale.

```python
hist_frames = []
for feature in numeric_features:
    counts, edges = np.histogram(df[feature], bins=40)
    hist_frames.append(
        pd.DataFrame(
            {
                "feature": feature,
                "bin_center": (edges[:-1] + edges[1:]) / 2,
                "count": counts,
            }
        )
    )
df_hist = pd.concat(hist_frames, ignore_index=True)

fig = px.bar(
    df_hist,
    x="bin_center",
    y="count",
    facet_col="feature",
    facet_col_wrap=3,
    facet_row_spacing=0.05,
    facet_col_spacing=0.06,
    height=1400,
    title="Distribution of numeric features (40 bins each)",
    template="simple_white",
)
fig.update_xaxes(matches=None, showticklabels=True, title_text="")
fig.update_yaxes(matches=None, showticklabels=True, title_text="")
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))
fig.update_layout(bargap=0)
fig.show()
```

## Target Variable Relationships

### Numeric Features vs Target Variable

Box plots compare the distribution of each feature for delayed and on-time deliveries. As with the histograms, we compute the box statistics ourselves (quartiles, and whiskers that end at the most extreme value within 1.5 IQR of the box) and pass only those numbers to Plotly. Outlier points beyond the whiskers are not drawn; the IQR outlier counts later in this notebook show how many there are.

```python
def box_stats(values):
    """Quartiles and 1.5 IQR whiskers, as a standard box plot draws them."""
    q1, median, q3 = values.quantile([0.25, 0.5, 0.75])
    iqr = q3 - q1
    lower = values[values >= q1 - 1.5 * iqr].min()
    upper = values[values <= q3 + 1.5 * iqr].max()
    return q1, median, q3, lower, upper


statuses = ["Delayed", "On time"]
n_cols = 3
n_rows = int(np.ceil(len(numeric_features) / n_cols))

fig = make_subplots(
    rows=n_rows,
    cols=n_cols,
    subplot_titles=numeric_features,
    vertical_spacing=0.05,
    horizontal_spacing=0.08,
)
for i, feature in enumerate(numeric_features):
    for status in statuses:
        q1, median, q3, lower, upper = box_stats(
            df.loc[df["classification_ontime"] == status, feature]
        )
        fig.add_trace(
            go.Box(
                x=[status],
                q1=[q1],
                median=[median],
                q3=[q3],
                lowerfence=[lower],
                upperfence=[upper],
                name=status,
                marker_color=px.colors.qualitative.D3[0],
                showlegend=False,
            ),
            row=i // n_cols + 1,
            col=i % n_cols + 1,
        )

fig.update_layout(
    height=1400,
    title="Numeric features by on-time status (outlier points not drawn)",
    template="simple_white",
)
fig.show()
```

Most weather features look almost the same for delayed and on-time deliveries. The clear exception is `total_items`: the median delayed delivery has about four times as many items on the truck as the median on-time delivery. We come back to this in the next notebook.

## Additional EDA Checks

The sections above cover the most common starting points for EDA. The checks below help surface issues that frequently affect downstream modeling but are easy to miss on a first pass.


### Duplicate Rows

Duplicate rows can inflate the apparent size of a dataset and bias both training and evaluation by leaking the same observations into multiple splits. We check for fully duplicated rows here.


```python
num_duplicates = df.duplicated().sum()
print(f"Number of fully duplicated rows: {num_duplicates}")
```

### Feature Cardinality

Cardinality is the number of unique values a feature takes. Very low cardinality (e.g., a constant column) provides no signal, while very high cardinality in a categorical column can make encoding tricky.


```python
cardinality = df.nunique().sort_values()
cardinality
```

### Constant or Near-Constant Columns

Columns where almost every row shares the same value rarely help a model and can be safely dropped.


```python
near_constant = [
    col
    for col in df.columns
    if df[col].value_counts(normalize=True, dropna=False).iloc[0] > 0.99
]
print("Near-constant columns (>99% same value):", near_constant)
```

`precipitation_rate` is 0 in every row, so it carries no information. It also breaks calculations that divide by a standard deviation, such as correlation. We leave it out of the models in the next notebooks.

### Outlier Detection (IQR Rule)

The interquartile range (IQR) rule flags values that fall below $Q_1 - 1.5 \cdot \text{IQR}$ or above $Q_3 + 1.5 \cdot \text{IQR}$ as potential outliers. Outliers are not always errors (extreme weather days, for example, may legitimately drive delays), but it is useful to know how many exist per feature.


```python
def count_iqr_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return ((series < lower) | (series > upper)).sum()


outlier_counts = {
    feature: int(count_iqr_outliers(df[feature])) for feature in numeric_features
}
pd.Series(outlier_counts, name="outlier_count").sort_values(ascending=False)
```

### Skewness

Skewness measures the asymmetry of a distribution. Highly skewed numeric features (|skew| > 1) often benefit from a transformation such as a log or square-root transform before being passed to linear models.


```python
skewness = df[numeric_features].skew().sort_values(ascending=False)
skewness
```

### Top Correlations with the Target

Encoding the target as 0/1 lets us rank numeric features by their (linear) correlation with on-time delivery. This is a quick way to spot the strongest individual predictors before building a model, though gradient-boosted trees can also capture non-linear effects that a correlation coefficient will miss.


```python
target_numeric = (df["classification_ontime"] == "On time").astype(int)

# A constant column has zero variance, so its correlation is undefined. Skip it.
varying_features = [col for col in numeric_features if df[col].nunique() > 1]

target_corr = (
    df[varying_features]
    .apply(lambda col: col.corr(target_numeric))
    .sort_values(key=lambda s: s.abs(), ascending=False)
)
target_corr
```

- The target is moderately imbalanced: about 58% of deliveries are "Delayed" and 42% are "On time".
- `precipitation_rate` is constant (always 0) and should be dropped. `zipcode` is an identifier stored as a number.
